```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):

    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

```python
# create Dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1]) # no. of input in one row

# call model for forward pass
# model.forward(features) # we can use this but
model(features) # pytorch recommend this , # here  i sent all 10 rows corresponding to which i got y_pred for each row
```

# making this NN complex

```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(3,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):

    out = self.linear1(features)
    out = self.relu(out)
    out = self.linear2(out)
    out = self.sigmoid(out)

    return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass
model(features)
```

```python
# show model weights
model.linear1.weight
```

```python
model.linear2.weight
```

```python
model.linear1.bias
```

```python
model.linear2.bias
```

```python
from torchinfo import summary
summary(model, input_size=(10,5))
```

### another method: sequential containers

```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3,1),
        nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)

    return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass

model(features)
```

```python
# show model weights and bias
print(model.network[0].weight)
print(model.network[0].bias)
print(model.network[2].weight)
print(model.network[2].bias)
```

```python
from torchinfo import summary
summary(model, input_size=(10,5))
```
